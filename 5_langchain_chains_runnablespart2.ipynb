```python
from abc import ABC, abstractmethod #ABC-asbsract base classes
```

```python
# Runnable is an abstract base class (ABC) that defines a common interface for all runnable components. 
# Subclasses must provide an implementation of the abstract invoke() method.

class Runnable(ABC):
  @abstractmethod
  def invoke(input_data):
    pass
```

```python
import random

class NakliLLM(Runnable):
  def __init__(self):
    print("llm created")

  # The invoke() method performs the same task as predict(), generating a response based on the provided prompt. 
  # It serves as the standard execution method defined by the Runnable interface.
  def invoke(self, prompt):
    response_list = [
        "delhi is the capital of india",
        "ipl stands for indian priemere league",
        "ai is artificial intelligence"
    ]
    return {"response": random.choice(response_list)}

  def predict(self, prompt):
    response_list = [
        "delhi is the capital of india",
        "ipl stands for indian priemere league",
        "ai is artificial intelligence"
    ]
    return {"response": random.choice(response_list)}
```

```python
class NakliPromptTemplate(Runnable):
  def __init__(self, t, i):
    self.template = t
    self.input_variables= i

  def invoke(self, input_dict):
    return self.template.format(**input_dict)

  def formate(self, input_dict):
    return self.template.format(**input_dict)
```

```python

class StrOutputParser(Runnable):
  def __init__(self):
    pass
    # No attributes are required during object initialization.
    # The class only needs an invoke() method to perform the parsing.

  def invoke(self, input):
    return input["response"]  # Extract the response value from the dictionary returned by NakliLLM.
    # Since NakliLLM returns a dictionary with "response" as its key, StrOutputParser extracts and returns its corresponding value.
```

```python

#Runnable Primitive
class RunnableConnector(Runnable):
  # Initialize the connector with a list of runnable components to be connected.
  # Each runnable's output will be passed as input to the next runnable.
  def __init__(self, list_of_runnables):
    self.runnable_list = list_of_runnables

  def invoke(self, input_data):
    # Sequentially execute each runnable in the list.
    for r in self.runnable_list:
      # Invoke the current runnable with the previous runnable's output.
      # The returned output is reassigned to input_data and passed as input to the next runnable during the subsequent iteration.
      input_data = r.invoke(input_data)

    # Return the final output after all runnables have been executed.
    return input_data
```

```python
prompt = NakliPromptTemplate(
    t = "write a {length} poem about {topic}",
    i = ['length','topic']
)
```

```python
llm = NakliLLM()
```

```python
parser = StrOutputParser()
```

```python
chain = RunnableConnector([prompt, llm, parser])
```

```python
chain.invoke({'length':'short', 'topic':'india'})
```

```python

template2 = NakliPromptTemplate(
    t = "explain the following joke {response}",
    i = ['response']
)
# The input variable is named "response" because the previous chain's LLM returns a dictionary containing the generated output under the "response" key.
# By using the same key in template2, the output of the previous chain can be passed directly as input to this chain without requiring key conversion.
```

```python
chain1= RunnableConnector([template1, llm])
```

```python
final_chain = RunnableConnector([chain1, chain2])
```

```python
final_chain.invoke({'topic': 'black hole'})
```
