```python
import random
class NakliLLM:
  def __init__(self):
    print("llm created")

  def predict(self, prompt):
    #since it is a fake llm method so let's just write dummy responses i.e responses irrespective of prompt
    response_list=[
        "delhi is the capital of india",
        "ipl stands for indian priemere league",
        "ai is artificial intelligence"
    ]
    return {"response": random.choice(response_list)}
```

```python
class NakliPromptTemplate:
  def __init__(self, t, i):
    self.template = t
    self.input_variables= i

  def formate(self, input_dict):
    return self.template.format(**input_dict)
```

```python
temp = NakliPromptTemplate(
    t = "write a {length} poem about {topic}",
    i = ['length','topic']
)
```

```python
pt = temp.formate({'length':'short', 'topic':'india'})
```

```python
llm = NakliLLM()
```

```python
llm.predict(pt)
```

```python
# A simplified implementation of the LLMChain approach, introduced to
# provide better compatibility between prompt templates and language models.

class NakliLLMChain:
  def __init__(self, l, p):
    self.llm = l
    self.prompt = p  # Store the PromptTemplate object in the prompt attribute.

  def run(self, input_dict):
    prmpt = self.prompt.formate(input_dict)  # Format the prompt using the provided input dictionary.
    result = self.llm.predict(prmpt)  # Pass the formatted prompt to the language model.
    return result['response']  # Extract and return the model's response.
```

```python
temp = NakliPromptTemplate(
    t = "write a {length} poem about {topic}",
    i = ['length','topic']
)
```

```python
llm = NakliLLM()
```

```python
chain = NakliLLMChain(llm, temp) 
```

```python
# With LLMChain, you don't need to explicitly call the prompt's format()
# and the LLM's predict() methods separately. The chain handles both operations internally through a single run() call.
chain.run({'length': 'short', 'topic': 'india'})
```
