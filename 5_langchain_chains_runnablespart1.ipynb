```python
class NakliPromptTemplate:
  def __init__(self, t, i):
    self.template = t
    self.input_variables= i

  def formate(self, input_dict):
    return self.template.format(**input_dict)
```

```python
temp = NakliPromptTemplate(
    t = "write a {length} poem about {topic}",
    i = ['length','topic']
)
#when we write like this Python looks at your __init__: i.e def __init__(self, t, i):"I need arguments named t and i, but say you give arguments named x and y here" then typeerror occurs
#But there's an important exception: positional arguments. You can use completely different variable names if you don't use keywords.say,
# temp = NakliPromptTemplate(
#     "write the poem about {topic}",
#     ["topic"]
# )
# This works:
# 1st value → t
# 2nd value → i because Python matches them by position.

# i.e
# You can absolutely do:
# x = "write the poem about {topic}"
# y = ["topic"]
# temp = NakliPromptTemplate(x, y)        this works!
```

```python
#now say for better compatibilty llmchain method was introduced:
class NakliLLMChain:
  def __init__(self, l, p):
    self.llm = l
    self.prompt = p #p parameter which is temp object get's stored in naklillmchain's obj's prompt attribute

  def run(self, input_dict):
    prmpt = self.prompt.formate(input_dict) #temp is obj of NakliPromptTemplate which has formate method not format so u just pass input_dict as parameter not **input_dict as u do in format
    result = self.llm.predict(prmpt)
    return result['response']
```

```python
temp = NakliPromptTemplate(
    t = "write a {length} poem about {topic}",
    i = ['length','topic']
)
```

```python
llm = NakliLLM()
```

```python
chain = NakliLLMChain(llm, temp) #u can see llm and temp are differnet names compared to l and p in init and still code worked...u know about positional arguements we read above so it works.
```

```python
chain.run({'length':'short', 'topic':'india'})
#u could see u don't have to seperately run predict and formate i.e individually run the methods of prompttemplate and llm.
```
